## Comparacao entre temperaturas

Temperaturas baixas devem produzir dominios alinhados; temperaturas altas, uma configuracao desordenada.

```python
painel = salvar_painel_temperaturas(
    saida / 'painel_temperaturas.png',
    [1.0, 1.5, 2.0, 2.27, 2.5, 3.0, 3.5],
    L=30, sweeps=300, J=1.0, h=0.0, rng=rng
)
painel
```
